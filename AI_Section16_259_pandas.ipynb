```python
mydataset ={
    'cars': ["BMW", "VOLVO", "Ford", "Tesla", "Ferrari", "Ford", "BMW"],
    'counts': [3, 7, 2, 8, 1, 9, 3]
}

print(mydataset)
print("======")

df = pd.DataFrame(mydataset)
print(df)
```

```python
#Methods that are widely used for cleaning the data
data_with_missing = {
    'cars': ["BMW", "Volvo", None, "Tesla", "Ferrari", None, "Ford"],
    'counts': [3, 7, None, 8, 1, 9, None]
}

df_missing = pd.DataFrame(data_with_missing)

print(df_missing)
```

```python
#counts missing values

df_missing.isnull().sum()
```

```python
#dropna -> removes rows where all values are missing

df_missing.dropna(how="all")
```

```python
#deletes the columns that have any missing value
df_missing.dropna(axis=1)
```

```python
#fillna() -> fills the missing values
df_missing.fillna("unknown")
```
